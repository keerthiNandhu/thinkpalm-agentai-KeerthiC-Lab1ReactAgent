```python
from forex_python.converter import CurrencyRates

# Tool
c = CurrencyRates()

# User query
user_query = "Convert 100 USD to INR"

print("User Query:")
print(user_query)

print("\nThought:")
print("I should use the currency converter tool to convert USD to INR.")

print("\nAction:")
print("Currency Converter Tool")

amount = 100
from_currency = "USD"
to_currency = "INR"

print("\nAction Input:")
print(f"{amount} {from_currency} to {to_currency}")

# Tool call
converted_amount = c.convert(from_currency, to_currency, amount)

print("\nObservation:")
print(f"{amount} {from_currency} = {converted_amount:.2f} {to_currency}")

print("\nFinal Answer:")
print(f"{amount} {from_currency} is equal to {converted_amount:.2f} {to_currency}")
```
